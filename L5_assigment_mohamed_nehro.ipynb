```python
class Student:
  def __init__(self, name, grades):
# TODO: ﺧﺰن اﻻﺳﻢ واﻟﺪرﺟﺎت ﻣﻦ اﻟﻤﻌﺎﻣﻼت — اﺳﺘﺒﺪل اﻟﻤﺆﻗﺖ
    self.name = name
    self.grades = grades

  def average(self):
# TODO: (ﻧﺎﺗﺞ ﻣﺆﻗﺖ) ﻟﻒ ﻋﻠﻰ اﻟﺪرﺟﺎت ورﺟﻊ اﻟﻤﺘﻮﺳﻂ
    total = 0
    for g in self.grades:
      total += g
    return total / len(self.grades)

s = Student("Sara", [85, 90, 76])
print(s.name)
print(s.average())
```

```python
assert s.name == "Sara", "اﻻﺳﻢ اﺗﺨﺰن ﻋﻠﻰ اﻟﻜﺎﺋﻦ"
assert abs(s.average() - 83.67) < 0.01, "ﻣﺘﻮﺳﻂ ﺳﺎرة"
t = Student("Omar", [70, 80])
assert t.name == "Omar", "ﻛﻞ ﻃﺎﻟﺐ ﻟﯿﮫ اﺳﻤﮫ"
assert abs(t.average() - 75.0) < 0.01, "ﻣﺘﻮﺳﻂ ﻋﻤﺮ"
print("A1: ﻣﻤﺘﺎز")
```

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        # TODO: ﺧﺰن اﺳﻢ اﻟﻤﺎﻟﻚ واﻟﺮﺻﯿﺪ
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        # TODO: زود اﻟﺮﺻﯿﺪ ﺑﻤﺒﻠﻎ اﻹﯾﺪاع
        self.balance += amount

    def withdraw(self, amount):
        # TODO: ﻟﻮ اﻟﻤﺒﻠﻎ أﻛﺒﺮ ﻣﻦ اﻟﺮﺻﯿﺪ اﻃﺒﻊ رﺳﺎﻟﺔ واﻗﻒ — ﻏﯿﺮ ﻛﺪه اﺧﺼﻢ
        if amount > self.balance:
            print("Insufficient balance!")
        else:
            self.balance -= amount

acc = BankAccount("Omar", 100)
acc.deposit(50)
```

```python
acc = BankAccount("Omar", 100)
acc.deposit(50)
assert acc.balance == 150, "اﻹﯾﺪاع زود اﻟﺮﺻﯿﺪ"
acc.withdraw(40)
assert acc.balance == 110, "اﻟﺴﺤﺐ ﺧﺼﻢ"
acc.withdraw(500)
assert acc.balance == 110, "اﻟﺤﻤﺎﯾﺔ ﻣﻦ اﻟﺴﺤﺐ اﻟﺰاﯾﺪ"
assert BankAccount("Sara").balance == 0, "اﻟﺮﺻﯿﺪ اﻻﻓﺘﺮاﺿﻲ ﺻﻔﺮ"
print("A2: ﻣﻤﺘﺎز")
```

```python
from typing import Self
class Book:
  def __init__(self, title, pages):
    # TODO: ﺧﺰن اﻟﻌﻨﻮان وﻋﺪد اﻟﺼﻔﺤﺎت
    self.title = title
    self.pages = pages

  def __str__(self):
    # TODO: رﺟﻊ اﻟﻨﺺ ﺑﺎﻟﺸﻜﻞ ده — ﺷﻮف ﺑﻠﻮك اﻟﺘﺤﻘﻖ ﺗﺤﺖ
    return f"{self.title} — {self.pages} ﺻﻔﺤﺔ"

b = Book("Python", 350)
print(b)
```

```python
assert str(b) == "Python — 350 ﺻﻔﺤﺔ", "اﻟﻨﺺ ﺑﺎﻟﻈﺒﻂ زي اﻟﻤﻄﻠﻮب"
b2 = Book("Pandas", 500)
assert str(b2) == "Pandas — 500 ﺻﻔﺤﺔ", "ﻛﺘﺎب ﺗﺎﻧﻲ"
print("A3: ﻣﻤﺘﺎز")
```

```python
class Classroom:
    def __init__(self, room):
        # TODO: اﺳﻢ اﻟﻘﺎﻋﺔ + ﻗﺎﺋﻤﺔ ﻃﻼب ﻓﺎﺿﯿﺔ
        self.room = room
        self.students = []

    def add(self, student):
        # TODO: ﺿﯿﻒ اﻟﻄﺎﻟﺐ ﻟﻘﺎﺋﻤﺔ اﻟﻄﻼب
        self.students.append(student)

    def best(self):
        # TODO: (ﻟﻒ ﻋﻠﻰ اﻟﻘﺎﺋﻤﺔ) رﺟﻊ اﻟﻄﺎﻟﺐ اﻷﻋﻠﻰ ﺑﻤﺘﻮﺳﻂ اﻟﺪرﺟﺎت
        if not self.students:
            return None
        best_student = max(self.students, key=lambda s: s.average())
        return best_student

room = Classroom("A101")
room.add(Student("Sara", [85, 90, 76]))
```

```python
assert room.room == "A101", "اﺳﻢ اﻟﻘﺎﻋﺔ"
assert len(room.students) == 1, "أول ﻃﺎﻟﺐ اﺗﻀﺎف"
room.add(Student("Omar", [95, 95]))
assert len(room.students) == 2, "ﺗﺎﻧﻲ ﻃﺎﻟﺐ"
assert room.best().name == "Omar", "اﻷﻋﻠﻰ ﺑﻤﺘﻮﺳﻂ ﯾﻔﻮز"
print("A4: ﻣﻤﺘﺎز")
```

```python
class Counter:
    def __init__(self):
        self.n = 0

    def bump(self):
        self.n += 1
        return self.n

c1 = Counter()
c2 = Counter()
print(c1.bump(), c1.bump(), c2.bump())
```
